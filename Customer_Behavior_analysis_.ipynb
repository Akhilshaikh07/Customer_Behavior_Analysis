```python
data.columns = data.columns.str.lower()
data.columns = data.columns.str.replace(" ","_")
data.columns = data.columns.str.strip()
```

```python
data.columns
```

```python
data = data.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
```

```python
data.columns
```

```python
data.isnull().sum()
```

```python
data["review_rating"] = data.groupby("category")["review_rating"].transform(lambda x: x.fillna(x.median()))
```

```python
data.isnull().sum()
```

```python
# Create column age group
labels = ["young","adult","midle_age","senior"]

data["age_group"] = pd.qcut(data["age"],q=4,labels = labels)
```

```python
print(data.head(10))
```

```python
data[["age","age_group"]]
```

```python
# create column purchase frequancy days

frequency_mapping = {
   " fortnightly": 14,
   "weekly": 7,
   "monthly": 30,
   "quarterly": 90,
   "big-weekly": 14,
   "Annually": 365,
   "every 3 month": 90
}

data["purchase_amount_days"] = data["frequency_of_purchases"].map(frequency_mapping)
```
